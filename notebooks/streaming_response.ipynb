```python
import sys
sys.path.append(r"G:\Projects\agentic-ai\LangGraph-Humanitarian-Report-Agent")

from agent.graph import app as aga



question = "What is the rice price in Bihar?"

import asyncio

async def ask(question):
    result= await aga.ainvoke({"question":question})
    return result

await ask(question)
```

```python
async def ask(question):
    async for chunk in aga.astream(
        {"question": question},
        stream_mode="values"
    ):
        keys = chunk.keys()
        print(f"State keys: {list(keys)}")
        print("---")

await ask(question)
```
